```sql
/****** Object:  View [dbo].[v_A01_result_list]    Script Date: 2026/07/27 18:32:55 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO







create view [dbo].[v_A01_result_list]
as
SELECT CONVERT(VARCHAR(100), CASE
                               WHEN Datename(hour, h.finish_date) < 8 THEN
                               Dateadd(day, -1, h.finish_date)
                               ELSE h.finish_date
                             END, 23)                     [生成实际日YMD]
       ,isnull(h.code,a.cd)                                         AS cd
	  --  ,h.code                                     AS cd
       ,m.goods_name                                      商品名-- 
       ,isnull(h.makenumber,a.[no])                                      作番
       ,h.productionquantity                              完工数量
       ,m_plan.worklinecode                               生产线
       ,ml.line_name                                      生产线名
       ,m_plan.destinationcode                            向先
       ,(SELECT CONVERT(VARCHAR(10), [idx]) + '|'
           FROM [m_picture_chk]
          WHERE a.[ck_id] = [chk_id]
         FOR xml path(''))                                图片
       --                pics.pics                                         AS 图片
       ,CONVERT(VARCHAR(100), CASE
                                WHEN Datename(hour, a.chk_start_date) < 8 THEN
                                Dateadd(day, -1, a.chk_start_date)
                                ELSE a.chk_start_date
                              END, 23)                    检查日期
       ,CONVERT(VARCHAR(100), a.chk_start_date, 20)       AS 检查开始时间
       ,CONVERT(VARCHAR(100), a.chk_end_date, 20)         AS 检查终了时间
       ,Datediff(s, a.chk_start_date, a.chk_end_date)     AS 检查时长
       ,a.result                                          AS 检查结果
       ,a.chk_user + '(' + Isnull(b.user_name, '') + ')'  AS 检查者
       ,a.edit_user + '(' + Isnull(r.user_name, '') + ')' AS 修正者
       ,(SELECT mark + ','
           FROM t_check_ms
          WHERE Isnull(mark, '') <> ''
                AND a.[ck_id] = [ck_id]
         FOR xml path(''))                                备注
       ,a.department_cd                                   部门
       ,b.line_cd                                         检查者生产线
       ,CASE
          WHEN Datename(hour, a.chk_start_date) >= 8
               AND Datename(hour, a.chk_start_date) <= 16 THEN '2直'
          WHEN Datename(hour, a.chk_start_date) >= 2
               AND Datename(hour, a.chk_start_date) <= 7 THEN '1直'
          ELSE '3直'
        END                                               AS 班直
  FROM t_check a
       LEFT OUTER JOIN dbo.m_user b
                    ON a.chk_user = b.user_cd
       LEFT OUTER JOIN dbo.m_user r
                    ON a.edit_user = r.user_cd
       LEFT JOIN tb_completedata h
              ON h.code = a.cd_no_sign
                 AND h.makenumber = a.[no]
       LEFT JOIN [tcm_bianplan] m_plan
              ON h.code = Replace(m_plan.[productcode], '-', '')
                 AND h.makenumber = m_plan.[zuofan]
       LEFT JOIN m_goods_tmp m
              ON m.goods_cd = a.cd_no_sign
       LEFT JOIN m_line ml
              ON ml.line_cd = a.line_cd 




GO
```

# [dbo].[v_last_check]

```sql
/****** Object:  View [dbo].[v_last_check]    Script Date: 2026/07/27 18:32:55 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
CREATE VIEW [dbo].[v_last_check] AS
SELECT t.cd
	   ,REPLACE(t.cd,'-', '') o_cd
       ,t.[no]
       ,t.ck_id
       ,t.result
FROM   t_check AS t
INNER JOIN (SELECT cd
                    ,no
                    ,Max(ck_id) AS ck_id
            FROM   t_check
            GROUP  BY cd,no) AS maxt
ON t.ck_id = maxt.ck_id
GO
```

# [dbo].[v_A01_result_list_2026]

```sql
/****** Object:  View [dbo].[v_A01_result_list_2026]    Script Date: 2026/07/27 18:32:55 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO
create view [dbo].[v_A01_result_list_2026]
as
-- 1. 优先查出【能直接匹配上】的大多数数据（走索引，毫秒级）
SELECT 
    a.ck_id,
    a.cd_no_sign,
    a.[no],
    m1.worklinecode,
    m1.destinationcode
FROM t_check a
INNER JOIN tb_completedata h 
        ON h.code = a.cd_no_sign 
       AND h.makenumber = a.[no]
INNER JOIN [tcm_bianplan] m1 
        ON h.code = m1.[productcode] 
       AND h.makenumber = m1.[zuofan]

UNION ALL

-- 2. 只查【直连连不上】的极少数数据，再用 REPLACE 补漏
SELECT 
    a.ck_id,
    a.cd_no_sign,
    a.[no],
    m2.worklinecode,
    m2.destinationcode
FROM t_check a
INNER JOIN tb_completedata h 
        ON h.code = a.cd_no_sign 
       AND h.makenumber = a.[no]
INNER JOIN [tcm_bianplan] m2 
        ON h.code = REPLACE(m2.[productcode], '-', '') 
       AND h.makenumber = m2.[zuofan]
-- 关键条件：排除掉已经在第一个查询里匹配成功的组合
WHERE NOT EXISTS (
    SELECT 1 
    FROM [tcm_bianplan] sub 
    WHERE h.code = sub.[productcode] 
      AND h.makenumber = sub.[zuofan]
);
GO
```

# [dbo].[v_A05_check_scx]

```sql
/****** Object:  View [dbo].[v_A04_check_result]    Script Date: 2026/07/27 18:32:55 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO

CREATE VIEW [dbo].[v_A04_check_result]
AS
SELECT a.ck_id id
,CONVERT(VARCHAR(100), a.chk_start_date, 23)      生成实际日
       ,a.cd                                            cd
       ,g.[goods_name]                                  商品名
       ,a.no                                            AS 作番
       ,h.productionquantity                            完工数量
       ,a.line_cd                                       AS 生产线
       ,c.[line_name]                                   AS 生产线名
       ,d.destinationcode                               向先
       ,d.billno                                        纳期
       ,a.department_cd                                 AS 部门
       ,CONVERT(VARCHAR(100),    
	   case when DATENAME(HOUR, a.chk_start_date)<8 then dateadd(day,-1,a.chk_start_date)
	   else
	   a.chk_start_date
	   end
	   , 23)     检查日期
       ,a.chk_start_date                                AS 检查开始时间
       ,a.chk_end_date                                  AS 检查终了时间

       ,a.result                                        AS 检查结果
       ,
       --CASE 
       --	WHEN isnull(BB.result,'')='待' THEN '未检查' 
       --	WHEN isnull(tcr_kaisuu.cntid,0) < isnull(CC.cntid,0) THEN '未全检' 
       --	WHEN BB.result = 'NG' THEN '不合格未二检' 
       --	ELSE '' 
       --END jieguo, --原因
       -- tcr_kaisuu.cntid,CC.cntid,
       a.chk_user + '(' + Isnull(b.user_name, '') + ')' AS 检查者
	   ,a.edit_user + '(' + Isnull(r.user_name, '') + ')' AS 修正者
       ,remark.mark                                     备注
	   ,b.line_cd 检查者生产线
,a.[shared_ck_id] 继承结果ID
       ,Datediff(ss, a.chk_start_date, a.chk_end_date)  AS 检查时长
FROM   dbo.t_check a
       LEFT OUTER JOIN dbo.m_user b
                    ON a.chk_user = b.user_cd
       LEFT OUTER JOIN dbo.m_user r
                    ON a.edit_user = r.user_cd
       LEFT JOIN m_line c
              ON a.line_cd = c.line_cd
       LEFT JOIN [AvoidMiss_New].dbo.tb_completedata h
              ON h.makenumber = a.no
                 AND h.code = Replace(a.cd, '-', '')
       LEFT JOIN [dbo].[tcm_bianplan] d
              ON d.productcode = a.cd
                 AND d.zuofan = a.no
                 AND CONVERT(DATETIME, d.plandate, 120) = a.yotei_chk_date
                 AND d.[b2boderno] = a.[b2boderno]
                 AND d.[b2bindexno] = a.[b2bindexno]
       LEFT JOIN (SELECT [ck_id]
                         ,[mark] = Stuff ((SELECT '' + Isnull([mark], '')
                                           FROM   [dbo].[t_check_ms] AS t
                                           WHERE  t .[ck_id] =
                                                  [dbo].[t_check_ms].[ck_id]
                                           FOR xml path('')), 1, 0, '')
                  FROM   [dbo].[t_check_ms]
                  WHERE  [mark] IS NOT NULL
                         AND [mark] <> ''
                  GROUP  BY [ck_id]) remark
              ON remark.[ck_id] = a.[ck_id]
       LEFT JOIN (SELECT nom.cd
                         ,nom.no
                         ,nom.ck_id
                         ,nom.result
                  FROM   t_check AS nom
                         INNER JOIN (SELECT cd
                                            ,no
                                            ,Max(ck_id) AS ck_id
                                     FROM   t_check
                                     GROUP  BY cd
                                               ,no) AS maxt
                                 ON maxt.ck_id = nom.ck_id) AS BB
              ON BB.cd = a.cd
                 AND BB.no = a.no
       LEFT JOIN (SELECT Count(aaa.code) AS cntId
                         ,aaa.code
                         ,aaa.makenumber
                  FROM   [AvoidMiss_New].dbo.tb_completedata aaa
                  --INNER JOIN [AvoidMiss_New].dbo.tb_setallcheck bbb
                  --               ON aaa.code = bbb.code 
                  GROUP  BY aaa.code
                            ,aaa.makenumber) CC
              ON CC.makenumber = a.no
                 AND CC.code = a.cd
       LEFT JOIN (SELECT Count(ck_id) AS cntId
                         ,cd
                         ,no
                  FROM   t_check
                  GROUP  BY cd
                            ,no) tcr_kaisuu
              ON tcr_kaisuu.cd = a.cd
                 AND tcr_kaisuu.no = a.no
       LEFT JOIN [AvoidMiss_New].[dbo].[m_goods] g
              ON g.goods_cd = Replace(a.cd, '-', '') 
GO
```

```sql
/****** Object:  View [dbo].[v_A03_mi_check_result]    Script Date: 2026/07/27 18:32:55 ******/
SET ANSI_NULLS ON
GO
SET QUOTED_IDENTIFIER ON
GO





CREATE VIEW [dbo].[v_A03_mi_check_result]
AS
SELECT 
--ROW_NUMBER() OVER (ORDER BY h.code) AS RowNumber
		CONVERT(VARCHAR(100), CASE
                WHEN Datename(hour, h.finish_date) < 8
                THEN
                Dateadd(day, -1, h.finish_date)
                ELSE h.finish_date
                END, 23)					[生成实际日YMD]                -- YYYY-MM-DD
				,ml.department_cd 部门
                --,Replace(h.code, '-', '')                  AS cd --商品コード 
				,h.code                  AS cd --商品コード 
                ,m.goods_name                          商品名--     
                ,m_plan.cd_dealer + '—'
                 + m_plan.dealerabbreviation           AS 经销商
                ,m_plan.specialbookno                  AS 发注书
                ,h.makenumber                       作番--作番  
                ,h.productionquantity                  完工数量  --完工数量  
                ,m_plan.worklinecode                   生产线--生产线  
                ,m_plan.destinationcode                向先--向先  
                ,CONVERT(VARCHAR(100), h.pay_date, 23) 纳期日--纳期日 
                ,CASE
                   WHEN last_chk.cd IS NULL THEN '未检查'
                   WHEN cnt.cntId1 < cnt.cntId2 THEN '未全检'
                   WHEN last_chk.result = 'NG' THEN '不合格未二检'
                   ELSE ''
                 END                                   结果
				 ,h.finish_date [生成实际日时]
				 ,m_plan.[productcode] [新cd]
				 ,ml.line_name 生产线名
FROM  
tb_completedata h
       LEFT JOIN v_last_check last_chk
              ON h.code = last_chk.o_cd
                 AND  h.makenumber = last_chk.[no]
	   LEFT JOIN v_check_cp_cnt cnt
			 ON Replace(cnt.cd ,'-', '') = Replace(h.code ,'-', '')
			 AND cnt.[no] = h.makenumber
       LEFT JOIN  [tcm_bianplan] m_plan
              ON Replace(h.code ,'-', '') = Replace(m_plan.[productcode], '-', '')
                 AND h.makenumber = m_plan.[zuofan]
       LEFT JOIN m_goods_tmp m
              ON m.goods_cd = h.code
		LEFT JOIN m_line ml
		On ml.line_cd = m_plan.worklinecode
WHERE 

	--(last_chk.cd IS NULL Or cnt.cntId1 < cnt.cntId2 OR last_chk.result = 'NG')

    CASE
        WHEN last_chk.cd IS NULL THEN 'a'
        WHEN cnt.cntId1 < cnt.cntId2 THEN 'a'
        WHEN last_chk.result = 'NG' THEN 'a'
        ELSE ''
    END    <>''
--AND
--h.finish_date>'2021-08-19'





GO
```
